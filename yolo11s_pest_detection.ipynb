```python
from huggingface_hub import hf_hub_download

model_path = hf_hub_download(
    repo_id="underdogquality/yolo11s-pest-detection",
    filename="best.pt"
)

print(model_path)
```

```python
from ultralytics import YOLO

model = YOLO(model_path)

results = model.predict(
    source="/content/rice leaf roller.jpeg",
    imgsz=640,
    conf=0.25,
    save=True
)
```

```python
results
```

```python
result = results[0]

for box in result.boxes:
    class_id = int(box.cls[0])          # رقم الكلاس
    confidence = float(box.conf[0])     # نسبة الثقة
    insect_name = result.names[class_id]

    print(f"Insect: {insect_name}")
    print(f"Confidence: {confidence:.4f}")
```
